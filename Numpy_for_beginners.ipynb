# 3. Why was NumPy created?

Before NumPy, Python had some problems:

**Problem 1: Python lists were slow for calculations**

Example:

a = [1,2,3,4]
b = [5,6,7,8]


Adding these lists:

a + b


gives:

[1,2,3,4,5,6,7,8]


It joins lists instead of adding numbers.

Scientists needed:

[1+5, 2+6, 3+7, 4+8]


NumPy solved this.

**Problem 2: Large data processing was slow**

Scientists and engineers worked with:

Weather data
Physics calculations
Statistics
Engineering simulations

Normal Python was too slow.

NumPy introduced:

Faster arrays
Vector operations
Mathematical functions

**Problem 3: Memory usage**

Python lists store extra information for every element.

Example:

Python list:

Number → Object information → Memory


NumPy array:

Number Number Number Number

# 7. Important NumPy Functions

1. np.array()

Creates an array.

```python
np.array([1,2,3])
```

2. np.zeros()

Creates an array filled with zeros.

```python
np.zeros(5)
```

3. np.ones()

Creates an array filled with ones.

```python
np.ones(5)
```

4. np.arange()

Creates numbers in a range.

```python
np.arange(1,10,2)
```

5. np.linspace()

Creates evenly spaced numbers.

```python
np.linspace(0,10,5)
```

6. np.reshape()

Changes array shape.

```python
a = np.array([1,2,3,4,5,6])

a.reshape(2,3)
```

7. np.shape

Find array size.

```python
a.shape
```

8. np.sum()

Adds elements.

```python
a = np.array([1,2,3])

np.sum(a)
```

9. np.mean()

Find average.

```python
np.mean([10,20,30])
```

10. np.max()

Find maximum value.

```python
np.max([5,9,2])
```

11. np.min()

Find minimum value.

```python
np.min([5,9,2])
```

12. np.sort()

Sort values.

```python
np.sort([3,1,2])
```
