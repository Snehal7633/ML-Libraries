# 1. What is SciPy?

SciPy stands for Scientific Python.

It is a Python library used for advanced scientific and mathematical calculations.

In simple words:

NumPy gives Python powerful arrays, while SciPy gives Python advanced scientific tools that work on those arrays.

SciPy is mainly used in:

Engineering
Physics
Chemistry
Statistics
Data Science
Machine Learning
Signal Processing
Optimization problems

```python
#NumPy can calculate:
import numpy as np

a = np.array([1,2,3,4])

print(np.mean(a))
```

**SciPy can solve advanced problems like:**

1.Finding the best solution to an equation

2.Processing audio signals

3.Solving differential equations

4.Statistical analysis

# 3.Why was SciPy Created?

**Before SciPy, scientists used languages like:**

*MATLAB

*Mathematica

*Fortran

**These tools were powerful but had problems:**

**Problem 1:** Expensive software

MATLAB and similar tools required paid licenses.

Students and researchers needed free alternatives.

Solution:

SciPy provided free scientific computing using Python.

**Problem 2:** Python lacked advanced mathematics

Python could do basic calculations, but scientists needed:

Optimization

Statistics

Signal processing

Linear algebra

Integration

SciPy added these features.

**Problem 3:** Rewriting scientific algorithms

Researchers often had to write mathematical algorithms themselves.

SciPy provided tested and optimized functions.

# 7. scipy.linalg (Linear Algebra)

**Linear algebra deals with:**

*Matrices

*Vectors

*Equations

**Used in:**

*Machine learning

*Physics

*Engineering



**Matrix inverse**

Example:

```python
import numpy as np
from scipy import linalg

a = np.array([
    [1,2],
    [3,4]
])

result = linalg.inv(a)

print(result)
```

**Solving equations**

Equation:

x + 2y = 5
3x + 4y = 6

```python
from scipy.linalg import solve

a = [[1,2],
     [3,4]]

b = [5,6]

x = solve(a,b)

print(x)
```

# 8. scipy.optimize (Optimization)

**Optimization means:**

Finding the best possible answer.

Examples:

* Minimum cost
* Maximum profit
* Best parameters in ML

```python
from scipy.optimize import minimize

def function(x):
    return x**2 + 10

result = minimize(function, 0)

print(result)
```

9. scipy.stats (Statistics)

Used for:

* Probability

**Mean**

```python
from scipy import stats

data=[10,20,30,40]

print(stats.tmean(data))
```
